
Este notebook tem como objetivo ler o arquivo transacoes.csv e criar uma tabela delta chamada "sandbox.prcg.transacoes".

```python
%sql

SELECT * FROM read_files(
  "/Volumes/sandbox/prcg/prcg/transacao.csv",
  sep => ";"
)
LIMIT 5
```

```python
df = spark.read.format("csv")\
    .option("sep", ";")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/sandbox/prcg/prcg/transacao.csv")

df.limit(5).display()
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("sandbox.prcg.transacoes")
```
